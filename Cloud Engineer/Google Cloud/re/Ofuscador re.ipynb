```python
import re

# Archivos definidos en la práctica
fichero_entrada = "texto_original_2000_lineas_casos_raros_v2.txt"
fichero_salida = "mi_salida.txt"
fichero_esperado = "texto_ofuscado_2000_lineas_casos_raros_v2.txt"

# Lectura del fichero original
with open(fichero_entrada, 'r', encoding='utf-8') as f:
    lineas = f.readlines()

print(f"Total de líneas cargadas: {len(lineas)}")
```

```python
# 1. Nombre y apellidos: Tras "El cliente", soporta tildes, guiones y partículas [cite: 25-30]
regex_nombre = r'(?<=El cliente )([A-ZÁÉÍÓÚÑ][a-zñáéíóú]+(?:[\s-](?:de |del )?[A-ZÁÉÍÓÚÑ][a-zñáéíóú]+){1,3})'

# 2. Email: Dominio example.* (com, es, org), subdominios y alias [cite: 34-42]
regex_mail = r'\b[\w.+-]+@(?:[\w-]+\.)+example\.(?:com|es|org)\b'

# 3. Teléfono: Prefijo + o 00, extensiones ext. o x [cite: 43-54]
regex_telefono = r'(?:\+|00)\s?\(?\d{1,3}\)?(?:\s?[\s-]?\d{3,4}){2,3}(?:\s*(?:ext\.?|x)\s*\d+)?'

# 4. Cuenta: 4 formatos específicos, ignorando longitudes incorrectas [cite: 55-63, 81-85]
regex_cuenta = r'\b(?:[A-Z]{2}\s*[-\s/]\s*\d{10,12}|[A-Z]{3}-\d{8}-[A-Z]{2}|[A-Z]{2}\d{2}-\d{12})\b'

# 5. Dirección y Localidad: Estructura "domicilio <dir> de <loc>" [cite: 64-78]
regex_dir_loc = r'(domicilio\s+(?:en\s+)?)(?:c/|calle|avd\.|avda\.|avenida|Av\.|Avda)\.?\s+([^,]+?)\s+(de\s+)([A-ZÁÉÍÓÚÑ][a-zñáéíóú]+(?:\s+[A-ZÁÉÍÓÚÑ][a-zñáéíóú]+)*)'
```

```python
lineas_ofuscadas = []

for linea in lineas:
    # Reemplazos simples
    linea = re.sub(regex_mail, '<mail>', linea)
    linea = re.sub(regex_telefono, '<telefono>', linea)
    linea = re.sub(regex_cuenta, '<cuenta>', linea)
    linea = re.sub(regex_nombre, '<nombre_apellidos>', linea)
    
    # Reemplazo complejo: Dirección y Localidad
    def sustituir_dir_loc(match):
        # Grupo 1: conector inicial | Grupo 3: conector " de "
        return f"{match.group(1)}<direccion>{match.group(3)}<localidad>"
    
    linea = re.sub(regex_dir_loc, sustituir_dir_loc, linea)
    lineas_ofuscadas.append(linea)

# Escritura del fichero de salida solicitado
with open(fichero_salida, 'w', encoding='utf-8') as f:
    f.writelines(lineas_ofuscadas)

print(f"Fichero '{fichero_salida}' generado correctamente.")
```
