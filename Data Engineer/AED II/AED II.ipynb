```python
from sklearn.linear_model import LinearRegression
```

```python
x = df[['RM']] # Variable independiente
y = df['MEDV'] # Variable dependiente
model = LinearRegression() # Crear el modelo de regresión lineal
model.fit(x, y) # Ajustar el modelo
print("Pendiente:", model.coef_[0]) # Coeficiente de la variable independiente
print("Intersección:", model.intercept_) # Intersección con el eje Y
```

```python
plt.scatter(x,y) # Diagrama de dispersión de los datos
plt.plot(x, model.predict(x), color='red') # Línea de regresión
plt.title('Regresión Lineal: RM vs MEDV') # Título del gráfico
plt.xlabel('Numero medio de habitaciones (RM)') # Etiqueta del eje X
plt.ylabel('Precio (MEDV)') # Etiqueta del eje Y
plt.show() # Muestra el gráfico
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

```python
x = df.drop(columns=['MEDV']) # Características (todas las columnas excepto 'MEDV')
y = df['MEDV'] # Variable objetivo
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42) # Dividir en conjunto de entrenamiento y prueba
```

```python
model = LinearRegression() # Crear el modelo de regresión lineal
model.fit(x_train, y_train) # Entrenar el modelo
y_pred = model.predict(x_test) # Hacer predicciones
```

```python
print("MSE:", mean_squared_error(y_test, y_pred)) # Error cuadrático medio
print("R^2:", r2_score(y_test, y_pred)) # Coeficiente de determinación
```
