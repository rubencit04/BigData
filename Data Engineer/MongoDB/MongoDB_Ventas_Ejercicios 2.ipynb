## 2) Ejercicios (queries con `find`)

En cada ejercicio tienes **enunciado** y una **solución**.

> Nota: `find()` no calcula campos (como facturación). Para eso usamos `aggregate()`.


### Ejercicio 1 — Ventas de un producto concreto

**Enunciado:** listar todas las ventas del producto `Ratón`.


````python
# Busca todos los documentos donde el campo "producto" sea igual a "Ratón"
ventas_raton = ventas.find({"producto": "Ratón"})

print("Ventas de Ratón encontradas:")
for venta in ventas_raton:
    print(venta)
````

### Ejercicio 2 — Ventas con cantidad grande

**Enunciado:** listar ventas con `cantidad >= 5` (mostrar campos sin `_id`).


````python
# Filtra cantidad >= 5 y oculta el campo _id
ventas_cantidad = ventas.find(
    {"cantidad": {"$gte": 5}},  # Filtro: cantidad >= 5
    {"_id": 0}                  # Proyección: ocultar _id (mostrar el resto)
)

print("Ventas con cantidad >= 5:")
for venta in ventas_cantidad:
    print(venta)
````

### Ejercicio 3 — Ventas por rango de precio

**Enunciado:** ventas con `precio` entre 20 y 60 (incluidos).


````python
# Busca precio >= 20 Y precio <= 60
ventas_rango_precio = ventas.find({
    "precio": {
        "$gte": 20,
        "$lte": 60
    }
})

print("Ventas con precio entre 20 y 60:")
for venta in ventas_rango_precio:
    print(venta)
````

### Ejercicio 4 — Top 5 ventas más caras (por precio)

**Enunciado:** ordenar por `precio` descendente y traer 5.


````python
# Busca todo (filtro {}), ordena por precio DESC (-1) y limita a 5
top_ventas = ventas.find({}).sort("precio", -1).limit(5)

print("Top 5 ventas más caras:")
for venta in top_ventas:
    print(venta)
````

## 3) Ejercicios (agregación con `aggregate`)

Aquí calculamos métricas como **unidades** y **facturación**.



### Ejercicio 5 — Total de unidades vendidas (global)

**Enunciado:** sumar `cantidad` en toda la colección.


````python
pipeline = [
    {
        "$group": {
            "_id": None,              # Agrupar TODOS los documentos juntos
            "total_items": {"$sum": "$cantidad"}  # Sumar el valor del campo "cantidad"
        }
    }
]

# Ejecutamos la agregación
resultado = list(ventas.aggregate(pipeline))

print("Total de artículos en stock:", resultado[0]["total_items"])
````

### Ejercicio 6 — Facturación total (global)

**Enunciado:** calcular facturación total como suma de `cantidad * precio`.


````python

pipeline = [
    {
        "$group": {
            "_id": None,
            "facturacion_total": {
                "$sum": { "$multiply": ["$cantidad", "$precio"] }
            }
        }
    }
]

# Ejecutamos la agregación
resultado = list(ventas.aggregate(pipeline))

# Imprimimos el resultado (redondeado a 2 decimales para que se vea bien)
total = resultado[0]["facturacion_total"]
print(f"Facturación total: {total:.2f} €")
````

### Ejercicio 7 — Unidades vendidas por producto

**Enunciado:** agrupar por `producto` y sumar `cantidad`.


````python
pipeline = [
    {
        "$group": {
            "_id": "$producto",  # Agrupa por el nombre del producto
            "total_unidades": { "$sum": "$cantidad" } # Suma las cantidades de ese grupo
        }
    }
]

resultados = list(ventas.aggregate(pipeline))

print("Total de unidades vendidas por producto:")
for item in resultados:
    print(f"Producto: {item['_id']} | Cantidad: {item['total_unidades']}")
````

### Ejercicio 8 — Facturación por producto (ranking)

**Enunciado:** por producto, calcular suma de `cantidad * precio` y ordenar desc.


````python

pipeline = [
    {
        "$group": {
            "_id": "$producto",
            "facturacion_producto": {
                "$sum": { "$multiply": ["$cantidad", "$precio"] }
            }
        }
    },
    {
        "$sort": { "facturacion_producto": -1 } # Orden descendente (de mayor a menor)
    }
]

resultados = list(ventas.aggregate(pipeline))

print("Facturación por producto (ordenado):")
for item in resultados:
    # Usamos :.2f para mostrar solo 2 decimales
    print(f"Producto: {item['_id']:<20} | Total: {item['facturacion_producto']:.2f} €")
````

### Ejercicio 9 — Resumen por producto (nº ventas, unidades, precio medio, facturación)

**Enunciado:** por producto, calcular:
- `n_ventas`: número de documentos
- `unidades`: suma de cantidades
- `precio_medio`: media del precio
- `facturacion`: suma de importe


````python
pipeline = [
    {
        "$group": {
            "_id": "$producto",
            # Cuenta 1 por cada documento
            "n_ventas": { "$sum": 1 },
            # Suma el valor del campo cantidad
            "unidades": { "$sum": "$cantidad" },
            # Calcula el promedio del campo precio
            "precio_medio": { "$avg": "$precio" },
            # Multiplica y luego suma
            "facturacion": { "$sum": { "$multiply": ["$cantidad", "$precio"] } }
        }
    },
    # Opcional: Ordenamos por facturación descendente para ver lo más importante primero
    {
        "$sort": { "facturacion": -1 }
    }
]

resultados = list(ventas.aggregate(pipeline))

print(f"{'PRODUCTO':<20} | {'VENTAS':<6} | {'UNID.':<6} | {'PRECIO MED':<10} | {'FACTURACIÓN':<12}")
print("-" * 70)

for r in resultados:
    print(f"{r['_id']:<20} | {r['n_ventas']:<6} | {r['unidades']:<6} | {r['precio_medio']:<10.2f} | {r['facturacion']:<12.2f}")
````

### Ejercicio 10 — Histograma de precios (rangos) PUNTO EXTRA !!

**Enunciado:** crear un campo `rango_precio` y contar cuántas ventas hay en cada rango:
- `<10`
- `10-25`
- `25-50`
- `>=50`

> Se usa `$switch` (más portable que `$bucket` si tu entorno no lo soporta).


````python
pipeline = [
    {
        "$project": {
            # Mantenemos el _id si quieres, o no, pero aquí creamos el campo calculado
            "rango_precio": {
                "$switch": {
                    "branches": [
                        # Caso 1: Precio menor a 10
                        { "case": { "$lt": ["$precio", 10] }, "then": "< 10" },
                        # Caso 2: Precio menor a 25 (ya sabemos que es >= 10 por el caso anterior)
                        { "case": { "$lt": ["$precio", 25] }, "then": "10-25" },
                        # Caso 3: Precio menor a 50 (ya sabemos que es >= 25)
                        { "case": { "$lt": ["$precio", 50] }, "then": "25-50" }
                    ],
                    # Caso por defecto: Si no cumple ninguno de los anteriores (es >= 50)
                    "default": ">= 50"
                }
            }
        }
    },
    {
        "$group": {
            "_id": "$rango_precio",  # Agrupamos por la etiqueta que acabamos de crear
            "cantidad_ventas": { "$sum": 1 }
        }
    },
    # Opcional: Ordenamos para que la salida sea legible
    {
        "$sort": { "_id": 1 }
    }
]

resultados = list(ventas.aggregate(pipeline))

print("Distribución de ventas por rango de precio:")
for r in resultados:
    print(f"Rango: {r['_id']:<10} | Ventas: {r['cantidad_ventas']}")
````

- Ranking Top 3 por facturación:




````python
# Crea un índice en el campo "producto" para acelerar las búsquedas
# Nota: Usamos una lista de tuplas [(campo, dirección)]
ventas.create_index([("producto", ASCENDING)])

print("Índice creado correctamente sobre 'producto'.")
````

````python
pipeline = [
    # 1. Agrupar por producto y calcular total (cantidad * precio)
    {
        "$group": {
            "_id": "$producto",
            "total_facturado": { "$sum": { "$multiply": ["$cantidad", "$precio"] } }
        }
    },
    # 2. Ordenar por facturación de mayor a menor (-1)
    {
        "$sort": { "total_facturado": -1 }
    },
    # 3. Quedarse solo con los 3 primeros
    {
        "$limit": 3
    }
]

ranking = list(ventas.aggregate(pipeline))

print("--- TOP 3 PRODUCTOS POR FACTURACIÓN ---")
for i, item in enumerate(ranking, 1):
    print(f"{i}. {item['_id']}: {item['total_facturado']:.2f}")
````
