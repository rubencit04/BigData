```python
# 1. Pedir nombres y notas
nombres_t = input("Introduce 5 nombres separados por comas: ")
notas_t = input("Introduce 5 notas (0–100) separadas por comas: ")

nombres_l = nombres_t.split(',')
notas_l = notas_t.split(',')

# 2. Unir nombres y notas en pares
pares = list(zip(nombres_l, notas_l))

# 3. Ordenar por nota descendente
ranking = sorted(pares, key=lambda par: float(par[1]), reverse=True)

# 4. Mostrar top 1 y top 3
print("\n--- Ranking ---")
print("Top 1:", ranking[0])
print("Top 3:", ranking[:3])

# 5. Buscar nombre
nombre_buscar = input("\nIntroduce un nombre para buscar: ")
contador = nombres_l.count(nombre_buscar)
print(f"El nombre '{nombre_buscar}' aparece {contador} veces.")

# 6. Añadir pares nuevos
nuevo_nombre = input("\nIntroduce un nuevo nombre: ")
nueva_nota = input("Introduce su nota: ")
pares.append((nuevo_nombre, nueva_nota))

pares.insert(1, ("Lurdes", "58"))  # valor fijo de ejemplo

# 7. Reordenar y mostrar primero y último
ranking = sorted(pares, key=lambda par: float(par[1]), reverse=True)
print("\n--- Nuevo ranking ---")
print("Primer lugar:", ranking[0])
print("Último lugar:", ranking[-1])
```
