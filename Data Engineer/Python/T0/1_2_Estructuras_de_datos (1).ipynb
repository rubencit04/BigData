# Elementos de programación en Python
## Estructuras de datos

Ya hemos conocido la forma más básica de almacenar un valor o una cadena de caracteres a través de su asignación a una variable. En ocasiones, tenemos necesidad a lo largo de un código de guardar varios datos al mismo tiempo.

Python posee, además de los tipos de datos básicos, otros tipos de datos más complejos. Se trata de las **secuencias**, los **diccionarios** y los **conjuntos**. 


Estos tres tipos, pueden almacenar colecciones de datos de diversos tipos y se
diferencian por su sintaxis y por la forma en la cual los datos pueden ser manipulados.

Vamos a repasar cada uno de ellos.


### Secuencias

Una secuencia es un listado unidimensional y ordenado de valores que pueden ser de cualquier tipo, incluso otras estructuras de datos anidadas. Existen tres tipos bien diferenciados

<ul>
<li>Tuplas: listados inmutables.</li>
<li>Listas: listados mutables.</li>
<li>Cadenas de caracteres: inmutables y cuyos elementos son siempre caracteres.</li>
</ul>

Casi todas las operaciones disponibles se pueden aplicar sobre cualquier tipo de secuencia (excepto las que implican modificación que, lógicamente, sólo se pueden aplicar sobre listas).

Es relevante saber que los strings son secuencias, ya que influye en sus propiedades y las operaciones que se pueden realizar sobre las mismas.

##### <b>Creación de secuencias</b>

```python
tuple_1 = (1, 2, 3, 4)
list_1 = [1, 2, 3, 4]
str_1 = "esto es una cadena"

print(type(tuple_1))
print(type(list_1))
print(type(str_1))
print(tuple_1 == list_1) # Una tupla no es lo mismo que una lista
```

##### <b>Conversión/casting entre tipos de secuencia</b>

```python
tuple_1 = (1, 2, 3, 4)
list_1 = [1, 2, 3, 4]
str_1 = "esto es una cadena" # En Python 3.5 strings son Unicode por defecto

tuple_2 = tuple(list_1)
list_2 = list(tuple_1)
tuple_3 = tuple(str_1)

print(tuple_2)
print(list_2)
print(tuple_3)

print(type(tuple_2))
print(type(list_2))
print(type(tuple_3))
```

###### Creación de secuencias anidadas

Los elementos de las secuencias, tanto listas como tuplas, son hetereogéneos, así que es posible definir **secuencias que contienen otras secuencias**.
Esto nos puede servir para crear secuencias de varias dimensiones.

```python
tuple_1 = (1, 2, (3, 4), [1, 2, 3])
list_1 = [1, 2, 3, [4, 5], "prueba", (1, 2, 3)]
list_2 = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
print(tuple_1)
print(list_1)
print(list_2)
```

###### Concatenación de secuencias

```python
tuple_1 = (1, 2, 3, 4)
list_1 = [1, 2, 3, 4]
str_1 = "esto es una cadena"

tuple_2 = (5, 6, 7, 8)
list_2 = [5, 6, 7, 8]
str_2 = "esto es otra cadena"

tuple_3 = tuple_1 + tuple_2   # IMPORTANTE !!
list_3 = list_1 + list_2
str_3 = str_1 + str_2

print(tuple_3)
print(list_3)
print(str_3)
```

##### Slicing de secuencias

Extracción (mediante copia) de un conjunto ordenado de elementos de una secuencia.
Se realiza mediante la notación secuencia [inicio:final:salto]. Donde:
- inicio: Índice del primer elemento a extrar (en base 0). Si se omite se extrae desde el principio de la secuencia
- final: Índice del primero elemento que NO se extrae (en base 0). Si se omite se extrae hasta el final de la secuencia
- salto: Tamaño del salto a aplicar en la extracción. Si se omite se asume 1

```python
tuple_1 = (1, 2, 3, 4)
list_1 = [1, 2, 3, 4]
str_1 = "esto es una cadena"

print(tuple_1[:])
print(tuple_1[2:])
print(list_1[:3])
print(list_1[2:3])
print(str_1[:-3])
print(str_1[::2]) # Va saltando, 
print(str_1[::-1]) # Coloca una frase del reves y desde atras 
print(list_1[::-1])
```

##### Mezcla ordenada de secuencias

La función __zip__ permite crear _un objeto iterable_ a partir de los elementos de otras secuencias.

```python
## IMPORTANTE!! 
tuple_1 = (1, 2, 3, 4, 5)
list_1 = ["uno", "dos", "tres", "cuatro", "cinco"]
z = zip(tuple_1, list_1)
print(z)
print(list(z))
```

```python
import random 
lista1 = []
lista2 = []
lista_emparejada = []
for x in range(3):
    if x == 3:
        print("Lista llena")

    else:
        num_aleatorio = random.randint(0, 10)
        lista1.append(num_aleatorio)
        lista2.append(num_aleatorio + random.randint(0,5))

print(lista1)
print(lista2)
lista1.sort()
lista2.sort()
print("-----------------------------------------------")
print("Lista ordenada Ascendente: "+ str(lista1))
print("Lista ordenada Ascendente: "+ str(lista2))
print("-----------------------------------------------")
lista1.sort(reverse=True)
lista2.sort(reverse=True)
print("Lista ordenada Descendente: "+ str(lista1))
print("Lista ordenada Descendente: "+ str(lista2))
print("-----------------------------------------------")
lista_emparejada = zip(lista1, lista2)
for elemento1,elemento2 in lista_emparejada:
    print("Elemento 1: "+ str(elemento1) + " - Elemento 2: "+ str(elemento2))
```

También podemos hacer __unzip__ para separar una lista de tuplas en varias secuencias.

```python
s1 = [1, 2, 3 , 4]
s2 = ['primavera', 'verano', 'otoño', 'invierno' ]
z = zip(s1, s2)
c1, c2 = zip(*z) # UNZIP!
c1, c2
```

### Diccionarios (dict)

Un diccionario es una estructura que:
- Contiene un listado de pares **clave-valor**
- También se puede llamar array asociativo o *hash map*
- Sin orden


* La __clave__ de un diccionario puede ser cualquier variable de tipo inmutable:
    - cadenas, 
    - enteros, 
    - tuplas (con valores inmutables en sus miembros), etc. 

* Los __valores__ de un diccionario pueden ser de cualquier tipo: listas, cadenas, tuplas, otros diccionarios, objetos, etc.

##### Creación de un diccionario

```python
# Diccionario con contenido
dict_1 = {"clave_1": 1, "clave_2": "prueba"} 
print(dict_1)
print(bool(dict_1)) 

# Diccionario vacío
dict_2 = {}
print(dict_2)
print(bool(dict_2))
```

```python
# Este código, ¿va a fallar? ¿son todos los casos correctos?
dict_3 = {('a','b'):1, 'c':(4,5), 2:dict_1, [1,2]:'a'} # SOLUCION: [1,2]
print(dict_3)
```

```python
tuple_1 = (1, 2, 3, 4, 5)
list_1 = ["uno", "dos", "tres", "cuatro", "cinco"]
dict_1 = dict(zip(list_1, tuple_1))
print(dict_1)
```

##### Comprobación de si un elemento existe en el diccionario

Si se intenta acceder a una clave que no está en el diccionario (ejemplo: dict_1[0], el acceso falla.
Se puede acceder únicamente a las claves, no a los valores.

Para evitar estos errores, podemos usar la función **in**, que comprueba si un elemento está en el diccionario o utilizar el método **get**, que devuelve el valor None si la clave no está en el diccionario.


```python
dict_1 = {"clave_1": 1, "clave_2": "prueba"}
print("clave_1" in dict_1)
print(1 in dict_1)
print(dict_1.get("clave_1"))
print(dict_1.get("1"))
```

### Conjuntos (set)

Un conjunto es una secuencia:
<ul>
<li>Unidimensional</li>
<li>Mutable</li>
<li>De tamaño variable (al ser mutable es obvio)</li>
<li>Desordenada</li>
<li>Cuyos elementos son valores o secuencias</li>
<li>Cuyos elementos son únicos</li>
</ul>

##### Cración de un conjunto

```python
set_1 = {1, 2, 3, 4, 5}
print(set_1)
```

##### Conversión/casting a conjunto

```python
list_1 = [1, 2, 3, 4, 5]
set_1 = set(list_1)
print(type(set_1))

list_3 = [1, 2, 3, 4, 5]
set_3 = set(list_3)
print(type(set_3))
```

##### Unión de conjuntos

```python
set_1 = {1, 2, 3, 4, 5}
set_2 = {3, 4, 5, 6, 7}
print(set_1 | set_2)
```

##### Intersección de conjuntos

```python
set_1 = {1, 2, 3, 4, 5}
set_2 = {3, 4, 5, 6, 7}
print(set_1 & set_2) # EXTRAE SOLO AQUELLOS ELEMENTOS QUE TIENEN AMBOS SET EN COMUN
```

##### Diferencia de conjuntos

```python
set_1 = {1, 2, 3, 4, 5}
set_2 = {3, 4, 5, 6, 7}
print(set_1 - set_2)
print(set_2 - set_1)
```

##### Diferencia simétrica de conjuntos

```python
set_1 = {1, 2, 3, 4, 5}
set_2 = {3, 4, 5, 6, 7}
print(set_1 ^ set_2)
print(set_2 ^ set_1)
```

##### Inserción de elementos en un conjunto

```python
set_1 = {1, 2, 3, 4, 5}
set_1.add(6)
print(set_1)
```

##### Comprobación de existencia de un elemento en el conjunto

```python
set_1 = {1, 2, 3, 4, 5}
print(3 in set_1)
```

##### Comprobación de subconjunto

```python
set_1 = {1, 2, 3, 4, 5}
set_2 = {1, 2, 3}
print(set_1 >= set_2)
```

##### Comprobación de superconjunto

```python
set_1 = {1, 2, 3, 4, 5}
set_2 = {1, 2, 3}
print(set_2 <= set_1)
```

##### Eliminación de un elemento de un conjunto por valor

```python
set_1 = {1, 2, 3, 4, 4, 5, 6, 7}
set_1.remove(5)
set_1.discard(6) # ENTENDER BIEN LA DIFERENCIA ENTRE REMOVE Y DISCARD
print(set_1)
```

##### Eliminación de todos los elementos de un conjunto

```python
set_1 = {1, 2, 3, 4, 5}
set_1.clear()
print(set_1)
```
