# Elementos de programación en Python
## Estructuras de control

Las estructuras de control de flujo son las que realmente permiten la construcción de programas. Estas estructuras son idénticas en funcionamiento al resto de lenguajes de programación, aunque variará ligeramente la sintáxis de las mismas.

### Condicionales (if, elif, else)

Esta estructura de control nos permitirá elegir, en base a una condición dada, la ejecución de diferentes bloques de código.

El intérprete de Python ejecuta un programa ejecutando una instrucción cada vez.

    if <condition>:
        <do something>
    elif <condition>:
        <do other thing>
    else:
        <do other thing>

Recordar que en Python los bloques se delimitan por sangrado. 
* Cuando ponemos los dos puntos al final de la primera línea del condicional, todo lo que vaya a continuación con __un__ nivel de sangrado superior se considera dentro del condicional. 
* En cuanto escribimos la primera línea con un nivel de sangrado inferior, hemos cerrado el condicional. 

```python
precio_taxi = 10.00
precio_bus = 2.50
dinero_disponible = 1.50
```

```python
if precio_taxi < dinero_disponible:
    print('Me voy en taxi')
    
if precio_bus < dinero_disponible and precio_taxi > dinero_disponible:
    print("Me voy en bus")
    
if precio_bus > dinero_disponible:
    print('Me voy andando')
```

Esta forma de realizar expresiones condicionales es muy redundante y obliga a la utilización de condiciones demasiado complejas. Se podría reescribir el mismo programa.

```python
if precio_taxi < dinero_disponible:
    print('Me voy en taxi')
elif precio_bus < dinero_disponible:
    print('Me voy en bus')
else:
    print('Me voy andando')
```

A continuación se muestran casos concretos para demostrar el funcionamiento del sangrado

```python
x, y = 2 , 0
if x > y:
    print("x es mayor que y")
elif x==y:
    print("x sigue siendo mayor que y")
```

```python
if 1 < 0:
    print("1 es menor que 0")
print("Esto se ejecuta siempre")  # esto no está dentro del bloque y no se ejecuta 
                                                    # dentro del if
```

```python
if 1 < 0:
    print("1 es menor que 0")
     print("1 sigue siendo menor que 0")    # error de sangrado
```

### Bucle - for

Esta estructura de control nos permitirá llevar a cabo la ejecución de un bloque de código un número fijo de veces. Se realizará la tarea para todos los elementos contenidos en una secuencia (que pueden estar involucrados o no en el bloque de código).

Se utiliza para recorrer una colección completa de elementos (una tupla, una lista, un diccionario, etc ) :

    for <element> in <iterable_object>:
        <hacer algo...>
        
* Aquí el objeto `<iterable_object>` puede ser una lista, tupla, array, etc.
* El bucle se repite un número fijo de veces, que es la longitud de la colección de elementos.


```python
for i in range(0, 10):
    print("Hola mundo")
```

```python
for i in range(0, 10):
    print("Hola mundo: " + str(i))
```

```python
list_1 = ["uno", "dos", "tres", "cuatro", "cinco"]
for element in list_1:
    print(element)
```

A veces es muy cómodo el disponer al mismo tiempo del elemento sobre el que se itera y del índice que ocupa dentro del bucle. Para ello podemos utilizar la función <i>enumerate</i>.

```python
list_1 = ["uno", "dos", "tres", "cuatro", "cinco"]
for index, element in enumerate(list_1):
    print("Elemento en índice " + str(index) + ": " + element)
```

En ciertos casos nos puede interesar recorrer el elemento sobre el que se itera en sentido inverso. Para ello, usamos la función reversed

```python
list_1 = ["uno", "dos", "tres", "cuatro", "cinco"]
for (i, valor) in enumerate(reversed(list_1)):
    print('%d: %s' % (i, valor))  # ENTENDER BIEN ESTA LINEA
```

También es posible recorrernos una tupla de elementos:

```python
tupla = ((0, 1), (1, 2), (1, 3))
for x in tupla:
  print(sum(x))

for x in tupla:
  print(x[0] + x[1])  

  for (x, y) in tupla:
    print(x + y)
```

Un diccionario también es un elemento iterable:

```python
dic =  {1:'Lunes', 2:'Martes', 3:'Miercoles' }
# Python 3.5
for i in dic.items():
    print(i)
```

En el caso de los diccionarios podemos iterar también sobre las claves y sobre los valores.

```python
for i in dic.keys():    
    print(i)
```

```python
for i in dic.values():  
    print(i)
```

### Bucle - while

Esta estructura de control nos permitirá llevar a cabo la ejecución de un bloque de código mientras se cumpla una condición dada.

Los bucles __while__ repetirán las instrucciones anidadas en él mientras se cumpla una condición:

    while <condition>:
        <things to do>

* El número de iteraciones es variable, depende de la condición.
* Como en el caso de los condicionales, los bloques se separan por sangrado sin necesidad de instrucciones del tipo `end`.

```python
i = 0
while i < 10:
    print (i)
    i += 1
print("Estoy fuera del while")
```

### Interrupción de bucles - break y continue

Las sentencias break y continue permiten alterar el flujo normal de ejecución de un bucle. Más concretamente:

<ul>
<li>break: Corta completamente el flujo de ejecución del bucle y pasa el control a la sentencia posterior al bucle.</li>
<li>continue: Corta la ejecución de la iteración actual y pasa el control a la siguiente iteración del bucle. Es decir, permite saltar de una iteración a otra</li>
</ul>

```python
list_1 = ["uno", "dos", "tres", "cuatro", "cinco"]
for element in list_1:
    if element == "cuatro":
        break
    print (element)
```

```python
list_1 = ["uno", "dos", "tres", "cuatro", "cinco"]
for element in list_1:
    if element == "cuatro":
        continue
    print (element)
```

### Comprensiones de lista

Dada la filosofía de Python de hacer el código más conciso, tenemos a nuestra disposición una estructura especial de control (semejante a los bucles) que facilita mucho la escritura de código. Imáginemos el siguiente bucle.

```python
list_1 = ["Miguel", "Juan", "Maria", "Manuel", "Rodrigo"]
list_M = []
for element in list_1:    
    if element[0] == "M":
        list_M.append(element.upper())  # ENTENDER BIEN ESTE COMANDO
print (list_M)
```

Podemos conseguir el mismo resultado con un código mucho más conciso.

```python
list_1 = ["Miguel", "Juan", "Maria", "Manuel", "Rodrigo"]
list_M = [element.upper() for element in list_1 if element[0] == "M"]
print (list_M)
```
