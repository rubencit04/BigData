🧠📘 Ejercicio avanzado: Mochila mágica de Link (modo experto)

El alumno deberá:

*Crear una lista desde texto con split().

*Añadir objetos si no están repetidos (append() + verificación).

*Eliminar por nombre (remove()) con control de errores.

*Eliminar el último (pop()), pero solo si hay más de 3 objetos.

*Contar cuántas veces aparece un objeto (count()).

*Mostrar una vista final ordenada alfabéticamente.

```python
mochila = input("Escribe objetos con ,")
mochila = mochila.split(",")
```

```python
nuevos_objetos_str = input("Escribe nuevos objetos para añadir con ,")
nuevos_objetos = [objeto.strip() for objeto in nuevos_objetos_str.split(",")]

for objeto_nuevo in nuevos_objetos:
    if objeto_nuevo not in mochila:
        mochila.append(objeto_nuevo)
        print(f"✅ '{objeto_nuevo}' ha sido añadido.")
    else:
        print(f"❌ '{objeto_nuevo}' ya estaba en la mochila, se ignora.")

print(f"\n Tu mochila final contiene: {mochila}")


```

```python
objetos_a_eliminar_str = input("\nEscribe el/los objeto(s) a eliminar, separados por comas: ")
lista_a_eliminar = [objeto.strip() for objeto in objetos_a_eliminar_str.split(",")]

for objeto in lista_a_eliminar:
    try:
        mochila.remove(objeto)
        print(f"✅ '{objeto}' ha sido eliminado de la mochila.")
    except ValueError:
        print(f"❌ '{objeto}' no estaba en la mochila, no se pudo eliminar.")

print(f"\n Tu mochila final contiene: {mochila}")
```

```python
print(mochila)
if len(mochila) > 3:
    objeto_eliminado = mochila.pop()
    print(f"\n Se ha eliminado el último objeto: '{objeto_eliminado}'")
    print(f" Tu mochila final contiene: {mochila}")
```

```python
objeto_a_contar = input("\n Escribe un objeto para contar cuántas veces aparece en la mochila: ")
veces = mochila.count(objeto_a_contar)
print(f" El objeto '{objeto_a_contar}' aparece {veces} veces en la mochila.")
```

```python
mochila_ordenada = sorted(mochila)
print(f"\n Tu mochila ordenada alfabéticamente: {mochila_ordenada}")
```
