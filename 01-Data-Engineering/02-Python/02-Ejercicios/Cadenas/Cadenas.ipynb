```python
cadena = "hola como estas"
def contar_vocales(cadena):
    vocales = "aeiouAEIOU"
    contador = 0
    for letra in cadena:
        if letra in vocales:
            contador += 1
    return contador
print("Vocales en la cadena: ", contar_vocales(cadena))
```

```python
palabra = "ojo rojo"
def palindromo(cadena):
    cadena = cadena.replace(" ","").lower()
    return cadena == cadena[::-1]
print("Es palindromo", palindromo(palabra))
```

```python
palabra = "hola chaval"
def palabra_mayuscula(palabra):
    palabra = palabra.upper()
    palabra = palabra.replace(" ", "_")
    return palabra
print("Palabra a mayuscula",palabra_mayuscula(palabra))
```

```python
palabra = "adios mendigo"
def contar_palabras(palabra):
    palabras =  palabra.split(" ")
    return len(palabras)
print("Palabras: ", contar_palabras(palabra))
```

```python
palabra= "Hola Caracola"
def extraer_iniciales(palabra):
    iniciales = ""
    palabras = palabra.split()
    for palabra in palabras:
        if palabra:  # Asegurarse de que la palabra no esté vacía
            iniciales += palabra[0].upper()
    return iniciales
print("Iniciales",extraer_iniciales(palabra))
```

```python
texto = "me cago en tus muertos"
prohibidas = ["cago", "muertos"]

def reemplazar_palabras_prohibidas(texto, prohibidas):
    for palabra in prohibidas:
        reemplazo = '*' * len(palabra)
        texto = texto.replace(palabra, reemplazo)
        texto = texto.replace(palabra.capitalize(), reemplazo)
        texto = texto.replace(palabra.upper(), reemplazo)
    return texto

print("Palabra reemplazada:", reemplazar_palabras_prohibidas(texto, prohibidas))
```

```python
palabra = "hola"
def invertir_palabra(palabra):
    return palabra[::-1]
print(invertir_palabra(palabra))
```

```python
nombre = "ruben"
apellido = "linares"
def formatear_nombre(nombre, apellido):
    return nombre.capitalize() + " " + apellido.capitalize()
print(formatear_nombre(nombre,apellido))
```

```python
texto = "Dios mio todo poderoso te amo"
palabra = "dios"
def buscar_palabra(texto, palabra):
    return palabra.lower() in texto.lower()
print(buscar_palabra(texto,palabra))
```

```python
palabras = ["java", "kotlin", "spring"]
def unir_lista_palabras(palabras):
    return ", ".join(palabras)
print(unir_lista_palabras(palabras))
```
