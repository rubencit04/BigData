### Ficheros en python

Abrir ficheros: modos y parámetros

Modos:

"r" leer, 
"w" escribir (trunca), 
"a" añadir, 
"x" crear si no existe.

Añade "b" para binario ("rb", "wb"), "t" para texto (por defecto).

Parámetros útiles en texto: encoding="utf-8", errors="ignore|replace", newline="" (CSV).

Usa SIEMPRE el context manager with para cerrar automáticamente.

```python
# Lectura de texto
with open("notas.txt", "r", encoding="utf-8") as f:
    notas = f.read()          # todo el fichero
```

print(notas)

```python
# 1) Escribir (crea o trunca)
with open("notas2.txt", "w", encoding="utf-8") as f:
    f.write("Primera línea\n")
    f.write("Segunda línea\n")

# 2) Volver a abrir en 'w' BORRA lo anterior y escribe de nuevo
with open("notas2.txt", "w", encoding="utf-8") as f:
    f.write("Contenido sobrescrito: ahora solo queda esta línea.\n")
```

```python
# Abrir en 'a' para AÑADIR al final
with open("notas2.txt", "a", encoding="utf-8") as f:
    f.write('holi\n')
    f.write('hola\n')
```

```python
import csv
# Leer filas como dicts
with open("datos.csv", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row['nombre'])

```

```python

# Sobreescribir
campos = ["id", "nombre", "importe"]
with open("datos.csv", "w", encoding="utf-8", newline="") as f:
    w = csv.DictWriter(f, fieldnames=campos)
    w.writeheader()
    w.writerow({"id": 1, "nombre": "juan", "importe": 5})
```

```python
# Append
campos = ["id", "nombre", "importe"]

# Filas a añadir (pueden venir de tu proceso ETL)
nuevas_filas = [
    {"id": 101, "nombre": "mar", "importe": 9.50},
    {"id": 102, "nombre": "nora",   "importe": 3.20},
]

# Abrimos en 'a' (append). newline="" evita líneas en blanco en Windows.
with open("datos.csv", "a", encoding="utf-8", newline="") as f:
    writer = csv.DictWriter(
        f, fieldnames=campos,
        extrasaction="ignore"  # ignora claves extra en los dicts
    )
    # Si el fichero está vacío, escribimos la cabecera una sola vez
    if f.tell() == 0:
        writer.writeheader()

    # Añadimos las nuevas filas
    writer.writerows(nuevas_filas)

print(f"Se añadieron {len(nuevas_filas)} filas al fichero csv")
```
