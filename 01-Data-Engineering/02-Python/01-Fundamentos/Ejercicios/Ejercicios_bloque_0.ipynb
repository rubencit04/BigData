#### 1. Escribe un código que diga de qué tipo es un input introducido por el usuario y lo convierta a todos los tipos posibles. Prueba a introducir un número, un texto y un número decimal, ¿qué ocurre?

```python
entrada = input("Introduce algo: ")

# Detección del tipo
def detectar_tipo(valor):
    try:
        int(valor)
        return "int"
    except:
        pass

    try:
        float(valor)
        return "float"
    except:
        pass

    return "string"

tipo = detectar_tipo(entrada)
print(f"El tipo detectado es: {tipo}")

# Conversiones
print("\nIntentando conversiones:")
try:
    print("A int:", int(entrada))
except:
    print("A int: no es posible")

try:
    print("A float:", float(entrada))
except:
    print("A float: no es posible")

print("A string:", str(entrada))
```

#### 2. Escribir un programa que reciba una lista de palabras separadas por coma y devuelva un diccionario en el que las claves sean las palabras introducidas y el valor el número de veces que se repite cada una de ellas en la lista. Emplea bucles tradicionales.

```python
entrada = input("Introduce palabras separadas por coma: ")
palabras = entrada.split(",")
conteo = {}

for palabra in palabras:
    palabra = palabra.strip() 
    if palabra in conteo:
        conteo[palabra] += 1
    else:
        conteo[palabra] = 1

print("Diccionario de conteo:", conteo)
```

#### 3. Escribir un programa que reciba una lista de palabras separadas por coma y devuelva la lista eliminando de la misma aquellos elementos que contengan 2 o más letras 'a'.

```python
entrada = input("Introduce palabras separadas por coma: ")

palabras = entrada.split(",")

filtradas = []

for palabra in palabras:
    palabra = palabra.strip() 
    if palabra.lower().count('a') < 2:
        filtradas.append(palabra)

print("Lista filtrada:", filtradas)
```

#### 4. Simplifica el mismo programa anterior con una compresión de listas.

```python
entrada = input("Introduce palabras separadas por coma: ")
palabras = [palabra.strip() for palabra in entrada.split(",")]
filtradas = [palabra for palabra in palabras if palabra.lower().count('a') < 2]
print("Lista filtrada:", filtradas)
```

#### 5. Escribir un programa que reciba un número que se utilizará como límite superior (incluido) de una secuencia numérica. El programa devolverá una lista de tuplas en las que cada tupla i-ésima contendra dos elementos: el elemento i-ésimo de la secuencia y un flag que indique si el número es o no primo.

```python
def es_primo(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

limite = int(input("Introduce el límite superior de la secuencia: "))

resultado = [(i, es_primo(i)) for i in range(1, limite + 1)]

print("Lista de tuplas:", resultado)
```

## 1.4 Funciones y módulos

#### 1. Escribe (y utiliza) una función que calcule el área de un círculo recibiendo como parámetro (opcional de valor por defecto 1) su radio. El módulo math de Python contiene una constante para PI.

```python
from math import math
def area_circulo(radio=1):
    return math.pi * radio ** 2
entrada = input("Introduce el radio del círculo (o deja vacío para radio=1): ")

if entrada.strip() == "":
    radio = 1
else:
    radio = float(entrada)

area = area_circulo(radio)

print(f"El área del círculo con radio {radio} es: {area:.2f}")

```

#### 2. Escribe una función que devuelva una lista con n (recibido como parámetro obligatorio) números aleatorios entre 0 y 1 generados con una distribución uniforme. El módulo random de Python contiene una función uniform.

```python
import random

def lista_aleatoria(n):
    numeros = []
    for _ in range(n):
        numeros.append(random.uniform(0, 1))
    return numeros

cantidad = int(input("Introduce cuántos números aleatorios quieres generar: "))
resultado = lista_aleatoria(cantidad)
print("Lista generada:", resultado)
```

```python
def contar_palabras(nombre_fichero):
    try:
        with open(nombre_fichero, "r", encoding="utf-8") as f:
            texto = f.read()
    except FileNotFoundError:
        print(f"El fichero '{nombre_fichero}' no existe.")
        return None

    palabras = texto.split()

    frecuencias = {}
    for palabra in palabras:
        palabra = palabra.strip().lower()
        if palabra in frecuencias:
            frecuencias[palabra] += 1
        else:
            frecuencias[palabra] = 1

    return frecuencias

nombre = input("Introduce el nombre del fichero: ")
resultado = contar_palabras(nombre)
if resultado:
    print("Frecuencias de palabras:", resultado)
```

#### 2. Escribe una función que reciba dos nombres de ficheros, el primero se utilizará para generar el diccionario del ejercicio anterior, el segundo para almacenar dicho diccionario en formato .tsv (tabulador = \t; salto de linea = \n)

```python
def contar_palabras(nombre_fichero):
    try:
        with open(nombre_fichero, "r", encoding="utf-8") as f:
            texto = f.read()
    except FileNotFoundError:
        print(f"El fichero '{nombre_fichero}' no existe.")
        return None

    palabras = texto.split()
    frecuencias = {}
    for palabra in palabras:
        palabra = palabra.strip().lower()
        if palabra in frecuencias:
            frecuencias[palabra] += 1
        else:
            frecuencias[palabra] = 1

    return frecuencias


def guardar_diccionario_tsv(fichero_entrada, fichero_salida):
    diccionario = contar_palabras(fichero_entrada)
    if diccionario is None:
        return

    try:
        with open(fichero_salida, "w", encoding="utf-8") as f:
            for palabra, frecuencia in diccionario.items():
                f.write(f"{palabra}\t{frecuencia}\n")
        print(f"Diccionario guardado en '{fichero_salida}' correctamente.")
    except Exception as e:
        print("Error al guardar el fichero:", e)

entrada = input("Introduce el nombre del fichero de entrada: ")
salida = input("Introduce el nombre del fichero de salida (.tsv): ")
guardar_diccionario_tsv(entrada, salida)
```
