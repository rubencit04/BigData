# Jupyter notebooks

Los notebooks son simplemente documentos JSON que contienen texto, código fuente, y metadatos. Podéis hacer la prueba abriéndolos con un editor de texto.

Los notebooks están formados por celdas. Cada celda puede contener código Python (marcado en el desplegable de arriba como "Code"), texto (marcado como "Markdown") o incluso código que no será ejecutado por el notebook ("Raw NBConvert"), por ejemplo en LaTeX para ser utilizado posteriormente cuando el documento sea convertido en un PDF.

Se puede guardar pulsando el botón de guardado o con el comando Ctrl + S

Los notebooks permiten ejecutar código de Python pero OJO, cerrar un notebook no es suficiente para detener su ejecución. Podremos ver qué codigos están corriendo en el Dashboard, que es la pantalla que se abre cuando iniciamos jupyter notebooks y a la que puedes acceder pulsando el icono de jupyter en la esquina superior izquierda.
Para ver los notebooks que están corriendo, ve a la pestaña "Running". Para detenerlos, selecciónalos y pulsa el botón "Shutdown".

Otros comandos interesantes en notebooks:
- Alt + Enter: insertar una nueva celda a continuación de la actual.
- Esc -> m: convierte la celda actual en Markdown
- Esc -> a: inserta una celda encima de la actual
- Esc -> d -> d: borra la celda actual
- Tab: muestra los atributos de un objeto
- Shift + Tab: muestra ayuda sobre cualquier función

## Código

  Cada celda de código se puede ejecutar de forma independiente. También podemos ejecutar un notebook entero de una sola vez, aunque hay que tener en cualquier caso, si una línea da un error no controlado, la ejecución del código se detendrá (si se han ejecutado varias celdas, la ejecución se parará igualmente).
Para ejecutarla:
- Pulsa el botón "Run" situado arriba
- Pusa Ctrl + Enter (se quedará seleccionada la celda en la que te encuentres)
- Pulsa Shift + Enter (se ejecutará y pasará a seleccionarse la siguiente celda)
Una celda sin ejecutar aparecerá con un "In [ ]:" a su izquierda.
Mientras se ejecuta, entre los corchetes veremos un asterisco, así [*].
Una vez ejecutados, los corchetes se rellenarán con un número.
Si el código de la celda está diseñado para devolver algún output, debajo del código aparecerá en rojo "Out [número del input]: " seguido de la salida del código.
Si una celda se ejecuta de nuevo, el número asociado cambiará.
