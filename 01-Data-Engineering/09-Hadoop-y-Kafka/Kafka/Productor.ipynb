```python
import pandas as pd
import json
import time
from kafka import KafkaProducer

producer = KafkaProducer(
    bootstrap_servers=['localhost:9092'], 
    value_serializer=lambda x: json.dumps(x).encode('utf-8')
)

# Cargar el 10% de los datos que guardamos en la Parte 1
df_streaming = pd.read_csv('dataset_10_streaming.csv')

print("🚀 Iniciando envío de datos...")

for index, row in df_streaming.iterrows():
    # Quitamos la 'y' real para hacer la predicción a ciegas
    mensaje = row.drop('y').to_dict()
    
    producer.send('bank_topic', value=mensaje)
    print(f"📤 Mensaje enviado (Fila {index})")
    
    # Pausa de 1.5 segundos
    time.sleep(1.5)
```
