```python

from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql import types as T

# Crear sesión Spark
spark = SparkSession.builder.appName("PySparkExercises").getOrCreate()

print("Spark version:", spark.version)
```

```python

# Crear DataFrame de ejemplo

data = [
(1,"Ana","Ventas",1200.0,34,"Madrid"),
(2,"Luis","IT",2200.0,29,"Barcelona"),
(3,"Marta","Ventas",1500.0,41,"Madrid"),
(4,"Pedro","RRHH",1800.0,37,"Sevilla"),
(5,"Lucia","IT",2400.0,31,"Valencia"),
(6,"Jorge","Marketing",1600.0,28,"Madrid"),
(7,"Elena","Ventas",1700.0,45,"Bilbao"),
(8,"Sergio","IT",2600.0,38,"Barcelona"),
(9,"Carmen","Marketing",1550.0,33,"Sevilla"),
(10,"Pablo","RRHH",1900.0,50,"Valencia"),
(11,"Irene","Ventas",2100.0,26,"Madrid"),
(12,"David","IT",2300.0,44,"Bilbao")
]

schema = ["id","nombre","departamento","salario","edad","ciudad"]

df = spark.createDataFrame(data,schema)

df.show()
```

## Ejercicio 1: Mostrar nombre y salario

```python

df.select("nombre", "salario").show()
```

## Ejercicio 2: Empleados con salario > 2000

```python

df.filter(F.col("salario") > 2000).show()
```

## Ejercicio 3: Número de empleados por departamento

```python


df.groupBy("departamento").count().show()
```

## Ejercicio 4: Salario medio por departamento

```python
df.groupBy("departamento").agg(F.avg("salario").alias("salario_medio")).show()
```

## Ejercicio 5: Crear columna salario anual

```python
df = df.withColumn("salario_anual", F.col("salario") * 12)
df.show()

```

## Ejercicio 6: Ordenar por salario descendente

```python
df.orderBy(F.col("salario").desc()).show()

```

## Ejercicio 7: Empleados de Madrid

```python
df.filter(F.col("ciudad") == "Madrid").show()
```

## Ejercicio 8: Edad mínima y máxima

```python
df.select(F.min("edad"), F.max("edad")).show()
```

## Ejercicio 9: Empleados por ciudad

```python
df.groupBy("ciudad").count().show()
```

## Ejercicio 10: Clasificación de salario

```python

df.withColumn("categoria", 
    F.when(F.col("salario") < 1600, "Bajo")
     .when((F.col("salario") >= 1600) & (F.col("salario") < 2300), "Medio")
     .otherwise("Alto")
).show()
```
