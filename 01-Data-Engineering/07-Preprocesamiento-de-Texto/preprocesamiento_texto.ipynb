```python
import nltk
nltk.download("twitter_samples")

[nltk]
```

```python
import pandas as pd
tweets = nltk.corpus.twitter_samples.strings()
```

```python
df = pd.DataFrame(tweets, columns=["tweet"])[:10]
```

```python
df.head()
```

```python
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\\S+", "", text)
    text = re.sub(r"<.*?>", "", text)
    text = re.sub(r"[^\\w\\s]", "", text)
    text = re.sub(r)
    return text

df["clean_text"] = df["tweet"].apply(clean_text) 
```

```python
import spacy
npl = spacy.load("en_core_web_sm")
```

```python
doc = npl("Los gatos estaban corriendo rápidamente")
```

```python
def tokenize_text(text):
    [token.lemma_ for token in doc if not token.is_stop]
df["clean_text"] = df["tweet"].apply(tokenize_text) 
```
