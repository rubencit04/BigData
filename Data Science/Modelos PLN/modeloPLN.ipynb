```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.linspace(0, 10, 50)
ruido = np.random.normal(0, 1, 50)
y = 2 * X + 0.5 + ruido
```

```python
w, b = 0.0, 0.0
alpha = 0.01
epochs = 1000
loss_history = []
```

```python
def mse(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)

def predict(X, w, b):
    return w * X + b
```

```python
#entrenamiento
for epoch in range(epochs):
    y_pred = predict(X, w, b)
error = y_pred - y
dw = (2/len(X)) * np.dot(X, error)
db = (2/len(X)) * np.sum(error)
w -= alpha * dw
b -= alpha * db
loss_history.append(mse(y, y_pred))

print(f"w = {w:.4f}, b = {b:.4f}")
```
