```python
import os
from google import genai
from google.genai.types import GenerateContentConfig
```

```python
GEMINI_API_KEY=''
MODEL_ID = "gemini-2.5-flash-lite"

client = genai.Client(api_key=GEMINI_API_KEY)
```

# Ejercicio 1: El experto en Pedagogía (Técnica de Asignación de Rol)

```python
prompt_ejercicio_1 = """
Actúa como un profesor de primaria experto en ciencias. 
Explícame qué es la fotosíntesis. 

Restricciones:
- Usa una analogía relacionada con una cocina o una fábrica para explicar el proceso.
- El lenguaje debe ser completamente apto y comprensible para niños de 8 años.
- Termina tu explicación obligatoriamente con una pregunta abierta para evaluar mi comprensión.
"""


response_1 = client.models.generate_content(
     model=MODEL_ID,
     contents=prompt_ejercicio_1
 )
print(response_1.text)
```

# Ejercicio 2: El estructurador de datos (Técnica de Formato de Salida)


```python
prompt_ejercicio_2 = """
A partir de la siguiente información desordenada sobre los beneficios de hacer ejercicio (salud cardiovascular, salud mental y fuerza muscular), organiza los datos estrictamente en una tabla comparativa.

Restricciones de formato:
- La tabla debe tener exactamente tres columnas con los siguientes encabezados: "Beneficio", "Explicación breve" y "Nivel de impacto (Alto/Medio)".
- NO incluyas ninguna introducción, saludo, conclusión ni texto adicional. 
- Devuelve ÚNICA Y EXCLUSIVAMENTE la tabla en formato Markdown.
"""


response_2 = client.models.generate_content(
    model=MODEL_ID,
              contents=prompt_ejercicio_2
 )
print(response_2.text)
```

# Ejercicio 3: Resolución por pasos (Cadena de Pensamiento / Chain of Thought)

```python
prompt_ejercicio_3 = """
Por favor, resuelve el siguiente problema lógico y matemático:
"Si tengo 3 cajas con 5 manzanas cada una y regalo 2 manzanas de cada caja, ¿cuántas me quedan?"

Instrucción obligatoria: 
Piensa paso a paso. Desglosa tu razonamiento de forma detallada y muestra cada cálculo intermedio antes de dar la respuesta final para evitar errores apresurados.
"""

response_3 = client.models.generate_content(
    model=MODEL_ID,
    contents=prompt_ejercicio_3
)
print(response_3.text)
```

# 🧪 Función para el Test de Estrés
Ejecuta el prompt 3 veces para asegurar que la IA no alucina y mantiene el formato.

```python
# Puedes usar esta pequeña función para validar cualquiera de los prompts anteriores
def test_de_estres(prompt_a_probar, repeticiones=3):
    print(f"--- Iniciando Test de Estrés ({repeticiones} iteraciones) ---")
    for i in range(repeticiones):
        print(f"\n--- Intento {i+1} ---")
        try:
            respuesta = client.models.generate_content(
                model=MODEL_ID,
                contents=prompt_a_probar
            )
            print(respuesta.text)
        except Exception as e:
            print(f"Error en la llamada a la API: {e}")
test_de_estres(prompt_ejercicio_2)
```
