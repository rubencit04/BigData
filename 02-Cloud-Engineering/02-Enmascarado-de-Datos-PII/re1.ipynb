```python
import re
```

```python
texto_1 = "Tengo 2 perros, 15 plantas y 300 fotos en el móvil."

# \d+ busca uno o más dígitos consecutivos
numeros = re.findall(r'\d+', texto_1)

print(f"Números encontrados: {numeros}")
```

```python
texto_2 = "Las reuniones son el 12/01/2026, 03/02/2026 y 7/3/2026."

# \d{1,2} busca 1 o 2 dígitos (día/mes)
# / busca la barra literal
# \d{4} busca 4 dígitos (año)
fechas = re.findall(r'\d{1,2}/\d{1,2}/\d{4}', texto_2)

print(f"Fechas encontradas: {fechas}")
```

```python
texto_3 = "Python3 es mejor que Java2, pero ambos son lenguajes."

# [a-zA-Z] busca letras mayúsculas y minúsculas
# El + indica "una o más veces"
# Esto separará "Python" de "3" y "Java" de "2"
palabras = re.findall(r'[a-zA-Z]+', texto_3)

print(f"Palabras encontradas: {palabras}")
```

```python
texto_4 = "Mis correos son prueba@gmail.com y usuario@empresa.com."

# [\w\.-]+ busca caracteres alfanuméricos, puntos o guiones
# @ busca la arroba literal
# \. busca el punto literal
emails = re.findall(r'[\w\.-]+@[\w\.-]+\.\w+', texto_4)

print(f"Emails encontrados: {emails}")
```

```python
texto_5 = "Clientes: 12345678A, 98765432Z, y un formato incorrecto 1234A."

# \b indica límite de palabra (para que sea exacto)
# \d{8} busca exactamente 8 dígitos
# [A-Z] busca una letra mayúscula
dnis = re.findall(r'\b\d{8}[A-Z]\b', texto_5)

print(f"DNIs válidos encontrados: {dnis}")
```

```python
texto_6 = "Los pedidos válidos son AB123, XY78 y ABC9999, pero no 123ABC"

# [A-Z]+ busca una o más letras mayúsculas al principio
# [0-9]+ busca uno o más números después
codigos = re.findall(r'\b[A-Z]+[0-9]+\b', texto_6)

print(f"Códigos encontrados: {codigos}")
```
